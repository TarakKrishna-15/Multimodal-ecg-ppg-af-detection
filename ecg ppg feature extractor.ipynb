```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ======================================
# LOAD FEATURES
# ======================================

ecg = np.load(
    "ecg_features2.npy"
)

ppg = np.load(
    "ppg_features2.npy"
)


print(
    "\nECG Shape:",
    ecg.shape
)

print(
    "PPG Shape:",
    ppg.shape
)


# ======================================
# CHECK SAMPLE COUNT
# ======================================

if ecg.shape[0] != ppg.shape[0]:

    raise ValueError(

        "\nECG and PPG sample count mismatch"

    )


# ======================================
# FEATURE FUSION
# ======================================

fused = np.concatenate(

    [

        ecg,

        ppg

    ],

    axis=1

)


print(
    "\nFusion Completed"
)

print(
    "Fused Shape:",
    fused.shape
)


# ======================================
# SAVE NPY
# ======================================

np.save(

    "fused_features2.npy",

    fused

)


print(
    "\nSaved:"
)

print(
    "fused_features2.npy"
)


# ======================================
# OPTIONAL CSV
# ======================================

df = pd.DataFrame(

    fused

)

df.to_csv(

    "fused_features2.csv",

    index=False

)


print(
    "fused_features2.csv"
)


# ======================================
# VERIFY
# ======================================

print(
    "\nNaN Count:",
    np.isnan(
        fused
    ).sum()
)

print(
    "\nMin:",
    np.min(
        fused
    )
)

print(
    "Max:",
    np.max(
        fused
    )
)

print(
    "Mean:",
    np.mean(
        fused
    )
)

print(
    "STD:",
    np.std(
        fused
    )
)


# ======================================
# VISUAL CHECK
# ======================================

plt.figure(

    figsize=(14,5)

)

plt.plot(

    fused[0]

)

plt.title(

    "Fusion Feature Vector"

)

plt.xlabel(

    "Feature Index"

)

plt.ylabel(

    "Activation"

)

plt.grid()

plt.show()


# ======================================
# FEATURE DISTRIBUTION
# ======================================

plt.figure(

    figsize=(8,5)

)

plt.hist(

    fused.flatten(),

    bins=50

)

plt.title(

    "Fusion Feature Distribution"

)

plt.xlabel(

    "Feature Value"

)

plt.ylabel(

    "Count"

)

plt.grid()

plt.show()


print(
    "\nFusion Step Completed"
)
```
